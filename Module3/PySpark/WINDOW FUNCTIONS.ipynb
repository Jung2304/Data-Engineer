```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window

df_sales_csv = (spark.read
                .format('csv')
                .option('header', True)
                .option('inferSchema', True)
                .load('/Workspace/PySpark/Dataset/BigMart_Sales.csv'))
```

# ROW_NUMBER

```python
# Assigns a sequential number to each row within a window.
df_sales_csv = df_sales_csv.withColumn('rowCol', row_number().over(Window.orderBy('Item_Identifier'))).display()

df_sales_csv.display()

"""
row_number()
      ↓
"Number the rows"

.over(...)
      ↓
"According to this window definition"

Window.orderBy(...)
      ↓
"Order them by Item_Identifier"
"""
```

# RANK

```python
df_sales_csv = df_sales_csv.withColumn('rank', rank().over(Window.orderBy('Item_Identifier')))

df_sales_csv.display()
```

# DENSE_RANK

```python
df_sales_csv = df_sales_csv.withColumn('dense_rank', dense_rank().over(Window.orderBy('Item_Identifier')))

df_sales_csv.display()
```

# CUMULATIVE SUM

### USING FRAME_CLAUSE

```python
df_sales_csv = df_sales_csv.withColumn('cumSum', sum('Item_MRP').over(Window
                                                                      .partitionBy('Item_Type')
                                                                      .orderBy('Item_Type')
                                                                      .rowsBetween(Window.unboundedPreceding, Window.currentRow)))

df_sales_csv.display()
"""
rowsBetween(Window.unboundedPreceding, Window.currentRow): Defines the window frame

first row ─────────────────→ current row
     ↑                           ↑
unboundedPreceding          currentRow

orderBy() → determines the order within the window
partitionBy() → creates separate windows/groups
rowsBetween() → determines which rows are included in the calculation
"""
```
