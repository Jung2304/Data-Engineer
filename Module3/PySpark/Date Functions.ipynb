```python
from pyspark.sql.functions import *
```

# Current_Date, Date_Add, Date_Sub

```python
df_sales_csv = (spark.read
                .format('csv')
                .option('header', True)
                .option('inferSchema', True)
                .load('/Workspace/PySpark/Dataset/BigMart_Sales.csv'))
```

```python
# Current_Date
df_sales_csv = df_sales_csv.withColumn('curr_date', current_date())

# Date_Add
df_sales_csv = df_sales_csv.withColumn('week_after', date_add('curr_date', 7))

# Date_Sub
df_sales_csv = df_sales_csv.withColumn('week_before', date_sub('curr_date', 7))
```

# Datediff

```python
df_sales_csv = df_sales_csv.withColumn('date_diff', date_diff('week_after', 'week_before'))
```

# Date_format

```python
df_sales_csv = df_sales_csv.withColumn("week_before", date_format("week_before", "dd-MM-yyyy"))
```
