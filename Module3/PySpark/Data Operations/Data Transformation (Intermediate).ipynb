```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

# UNION and UNION BYNAME

### Preparing Dataframes

```python
data1 = [
    ('1', 'kad'),
    ('2', 'sid')
]
schema1 = 'id STRING, name STRING'

df1 = spark.createDataFrame(data1, schema1)

data2 = [
    ('3', 'rahul'),
    ('4', 'jas')
]
schema2 = 'id STRING, name STRING'

df2 = spark.createDataFrame(data2, schema2)
```

## UNION (df1,df2)

```python
df1.union(df2).display()
```

## UNION BYNAME

```python
# Create a dataframe with name first, id later
data3 = [
    ('kad', '1'),
    ('sid', '2')
]
schema3 = "name STRING, id STRING"

df3 = spark.createDataFrame(data3, schema3)
```

```python
df2.unionByName(df3).display()
```

### initcap(), lower(), upper()

```python
# changes the first letter of every word in a text string to uppercase
df_sales_csv.select(initcap('Item_Type')).display()

df_sales_csv.select(lower('Item_Type')).display()

df_sales_csv.select(upper('Item_Type')).display()
```
