```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df = (spark.read
      .format('csv')
      .option('header', True)
      .option('inferSchema', True)
      .load('/Workspace/PySpark/Dataset/BigMart_Sales.csv'))
```

# IN THE CSV FORMAT

```python
(df.write
    .format('csv')
    .option('header', True)
    .save('/Workspace/PySpark/Data Storage/data.csv'))
```

# DATA WRITING MODES

## APPEND

```python
(df.write
 .format('csv')
 .mode('append')
 .option('header', True)
 .save('/Workspace/PySpark/Data Storage/data.csv'))
```

## OVERWRITE

```python
(df.write
 .format('csv')
 .mode('overwrite')
 .option('header', True)
 .save('/Workspace/PySpark/Data Storage/data.csv'))
```

## ERROR

```python
(df.write
 .format('csv')
 .mode('error')
 .option('header', True)
 .save('/Workspace/PySpark/Data Storage/data.csv'))
```

## IGNORE

```python
(df.write
 .format('csv')
 .mode('ignore')
 .option('header', True)
 .save('/Workspace/PySpark/Data Storage/data.csv'))
```
