```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df_sales_csv = (spark.read
                .format('csv')
                .option('header', True)
                .option('inferSchema', True)
                .load('/Workspace/PySpark/Dataset/BigMart_Sales.csv'))

display(df_sales_csv.limit(5))
```

# GROUP_BY

```python
df_eachItem = df_sales_csv.groupBy('Item_Type').agg(            # agg() tells Spark what calculation you want to perform on each group
    sum('Item_MRP').alias('total_mrp'),
    avg('Item_MRP').alias('avg_mrp'))      
```

```python
display(df_eachItem)
```

## GROUP_BY on 2 columns

```python
# Item_Type and Outlet_Size

df_eachItem = df_sales_csv.groupBy('Item_Type', 'Outlet_Size').agg(
    sum('Item_MRP').alias('Total_MRP'),
    avg('Item_MRP').alias('Avg_MRP'))
```

```python
display(df_eachItem)
```
