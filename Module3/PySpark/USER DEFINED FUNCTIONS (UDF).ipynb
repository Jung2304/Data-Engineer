```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df = (spark.read
      .format('csv')
      .option('header', True)
      .option('inferSchema', True)
      .load('/Workspace/PySpark/Dataset/BigMart_Sales.csv'))
```

### STEPS TO CREATE A UDF

```python
# STEP 1: Define
def my_func(x):
    return x*x

# STEP 2: Create a PySpark function (cohvert to PySpark)
my_udf = udf(my_func)

# STEP 3: Use the function
df = df.withColumn('myNewCol', my_udf("Item_MRP"))

df.display()
```
