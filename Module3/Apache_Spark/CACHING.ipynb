```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# Create first DataFrame
data1 = [
    (1, "Alice"),
    (2, "Bob"),
    (3, "Charlie"),
    (4, "David"),
    (5, "Eva")
]
df1 = spark.createDataFrame(data1, ["id", "name"])
```

## NOT CACHING DF1

```python
df1 = df1.withColumn('Flag', lit('Yes'))        # adds a new column called Flag to df1 and gives every row the value "Yes".

# lit: creates a literal constant value "Yes" for every row.
```

```python
display(df1)
```

```python
df2 = df1.filter(col('id') == 1)
```

```python
display(df2)

df2.explain(True)
```

### The line "Yes AS Flag" means that it recomputes df1 every time we run df2 (newly created).

## CACHING DF1


```python
df1.cache()
```

```python
df3 = df1.filter(col("id") == 1)
```

```python
display(df3)

df3.explain(True)
```
