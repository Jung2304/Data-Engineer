```python
from pyspark.sql.functions import * 
from pyspark.sql.types import *
```

```python
data = [
    ("Alice", 25, "New York"),
    ("Bob", 30, "San Francisco"),
    ("Charlie", 35, "Chicago")
]

schema = StructType([                                               # StructType represents the entire schema of the DataFrame.
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),                        # Each StructField defines one column.
    StructField("city", StringType(), True)
])

df = spark.createDataFrame(data, schema=schema)
```

## LAZY EVALUATON (TRANSFORMATION)

```python
# Filter (city = New York)
df_ny = df.filter(df.city == "New York")               
```

## ACTION

```python
# Action (show)
df_ny.show()
```

## Spark Query Plans

```python
## TRANSFORMATION 

# Filter (city = Chicago) and select only 'name' column
df_chicago = (
    df
    .filter(df.city == "Chicago")
    .select('name')
)
## ACTION
df_chicago.show()
```

```python
## To see the Query Plans of Spark
df_chicago.explain("extended")

"""
Spark performed logical analysis and optimization, and because our data is a tiny local dataset, the final physical execution was simplified to a LocalTableScan.
--> Use "extended" for more info
"""
```
