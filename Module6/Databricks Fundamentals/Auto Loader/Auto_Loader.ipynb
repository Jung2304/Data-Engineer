## Streaming Dataframe

```python
df_auto = (spark.readStream
                .format("cloudFiles")
                .option("cloudFiles.format", "parquet")
                .option("cloudFiles.schemaLocation", 
                        "abfss://alsink@datalakejung.dfs.core.windows.net/checkpoints/schema")
                .load("abfss://alsource@datalakejung.dfs.core.windows.net/"))    
```

```python
(df_auto.writeStream
    .format("delta")
    .option("checkpointLocation", 
            "abfss://alsink@datalakejung.dfs.core.windows.net/checkpoints/stream")
    .option("path",
            "abfss://alsink@datalakejung.dfs.core.windows.net/data")
    .trigger(availableNow=True)
    .start())
```

```python
dbutils.fs.rm(
    "abfss://alsink@datalakejung.dfs.core.windows.net/checkpoints/schema",
    True
)

dbutils.fs.rm(
    "abfss://alsink@datalakejung.dfs.core.windows.net/checkpoints/stream",
    True
)
```
