```python
from pyspark.sql import Row

data = [
    Row(id=1, name="Alice", amount=100),
    Row(id=2, name="Bob", amount=200),
    Row(id=3, name="Charlie", amount=300),
    Row(id=4, name="David", amount=400)
]

df = spark.createDataFrame(data)

display(df)

df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("sales.example.job_raw")
```
