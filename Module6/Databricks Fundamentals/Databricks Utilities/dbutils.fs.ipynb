## dbutils.fs: File operations

### dbutils.fs.ls: List files

```python
display(dbutils.fs.ls("abfss://source@datalakejung.dfs.core.windows.net/"))
```

### dbutils.fs.cp: Copy

```python
dbutils.fs.cp(
    "abfss://source@datalakejung.dfs.core.windows.net/Sales.csv",
    "abfss://source@datalakejung.dfs.core.windows.net/backup/sales_backup.csv"
)
```

### dbutils.fs.mv: Move

```python
dbutils.fs.mv(
    "abfss://source@datalakejung.dfs.core.windows.net/locations.csv",
    "abfss://source@datalakejung.dfs.core.windows.net/archive/locations.csv"
)
```

### dbutils.fs.rm: Delete

```python
dbutils.fs.rm("abfss://source@datalakejung.dfs.core.windows.net/archive/locations.csv")
```

### dbutils.fs.mkdirs: Create directory 

```python
dbutils.fs.mkdirs("abfss://source@datalakejung.dfs.core.windows.net/test_folder")
```
