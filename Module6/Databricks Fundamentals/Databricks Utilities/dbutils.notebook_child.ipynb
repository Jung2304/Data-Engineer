### Receive parameters and do some processing

```python
dbutils.widgets.text("name", "")
dbutils.widgets.text("age", "0")

name = dbutils.widgets.get("name")
age = int(dbutils.widgets.get("age"))

print(f"Hello {name}, welcome to my notebook!")
print(f"Your age is {age}")

double_age = age * 2

print(f"Double of your age is {double_age}")

dbutils.notebook.exit(str(double_age))
```
