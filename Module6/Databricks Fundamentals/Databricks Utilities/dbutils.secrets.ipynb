## dbutils.secrets
Used to retrieve credentials without putting them directly in notebooks.

#### dbutils.secrets.listScopes(): List available secret scopes

```python
dbutils.secrets.listScopes()
```

#### dbutils.secrets.list(): List available secret scopes

```python
dbutils.secrets.list("adls_scope")
```

#### dbutils.secrets.get(): Retrieve a secret value

```python
dbutils.secrets.get(
    scope = "adls_scope", 
    key = "databricks-masterclass-secret")
```

#### Use a secret to authenticate with ADLS Gen2

```python
storage_account = "datalakejung"
storage_key = dbutils.secrets.get(
        scope = "adls_scope",
        key = "databricks-masterclass-secret"
)

# Spark, whenever you access datalakejung.dfs.core.windows.net, use this storage key for authentication.
spark.conf.set(
    f"fs.azure.account.key.{storage_account}.dfs.core.windows.net",
    storage_key
)
```

Databricks environment does not allow setting this Spark configuration: fs.azure.account.key.datalakejung.dfs.core.windows.net  
This commonly happens in serverless / newer Databricks environments, where certain Spark configs are restricted.
