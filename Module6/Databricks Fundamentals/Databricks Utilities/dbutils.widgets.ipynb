## dbutils.widgets: Create parameters for notebooks.

### dbutils.widgets.text: 
Creates a text input widget at the top of a Databricks notebook, allowing you to pass parameters into your notebook.

```python
dbutils.widgets.text("p_name", "John Doe", "Name")

name = dbutils.widgets.get("p_name")

print(f"Welcome to my notebook, {name}")
```

### dbutils.widgets.dropdown
Creates a dropdown menu in a Databricks notebook, allowing the user to select one value from **predefined options**.

```python
# Choose the environment for production
dbutils.widgets.dropdown(
    "environment", 
    "dev",
    ["dev", "test", "prod"],
    "Environment"
)

environment = dbutils.widgets.get("environment")

print(f"Hey user! You have chose the environment to be {environment}!")
```
