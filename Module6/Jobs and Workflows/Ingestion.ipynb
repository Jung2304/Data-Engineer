```sql
%python

import json

dbutils.widgets.text("file_name", "")

# Convert the widget string back into a Python list
file_names = json.loads(
    dbutils.widgets.get("file_name")            # always returns a string --> need to convert to Python list
) 
```

```sql
%python

for file_name in file_names:    
    df = (spark.read
        .format("parquet")
        .load(f"/Volumes/db_jobs/default/raw_data/source/{file_name}.parquet"))           # dynamic
    
    df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"/Volumes/db_jobs/default/raw_data/sink/{file_name}")
```
