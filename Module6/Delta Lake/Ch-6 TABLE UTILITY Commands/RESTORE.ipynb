# SCENARIO
Accidentally merged corrupted data; we want to revert back to before the merge

### DATA VERSIONING

```sql
-- Check all the history versions of the table
DESCRIBE HISTORY delta.`/Volumes/datalake_jung/default/delta_volume/dml_sink/`
```

```sql
SELECT * 
FROM delta.`/Volumes/datalake_jung/default/delta_volume/dml_sink/`
```

### TIME TRAVELING (BY RESTORE)

```sql
-- Revert back to before the merge
RESTORE TABLE delta.`/Volumes/datalake_jung/default/delta_volume/dml_sink/`
TO VERSION AS OF 3 
```

```sql
SELECT * 
FROM delta.`/Volumes/datalake_jung/default/delta_volume/dml_sink/`
```

### TIME TRAVELING (BY SELECT)

```sql
SELECT *
FROM delta.`/Volumes/datalake_jung/default/delta_volume/dml_sink/`
VERSION AS OF 1
```
