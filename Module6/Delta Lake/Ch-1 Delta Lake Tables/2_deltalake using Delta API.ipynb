# CREATE DELTA TABLE

```sql
%python
from delta.tables import DeltaTable, IdentityGenerator

DeltaTable.createIfNotExists(spark) \
    .tableName("datalake_jung.default.first_delta_api") \
    .addColumn("id", "INT") \
    .addColumn("name", "STRING") \
    .execute() 
```

```sql
%python
# variable pointing to my Delta table 
path = "datalake_jung.default.first_delta_api"
first_delta_api_table = DeltaTable.forPath(spark,path)          # Python variable contains DeltaTable object
```

# GENERATED COLUMNS

```sql
%python
DeltaTable.createIfNotExists(spark) \
    .tableName("datalake_jung.default.second_delta_api") \
    .addColumn("id", "BIGINT", generatedAlwaysAs=IdentityGenerator(1,1)) \
    .addColumn("department", "STRING") \
    .addColumn("salary", "DECIMAL(18,2)") \
    .addColumn("salaryAfterTax", "DECIMAL(20,3)", generatedAlwaysAs="salary * 0.7") \
    .addColumn("first_name", "STRING") \
    .addColumn("last_name", "STRING") \
    .addColumn("full_name", "STRING", generatedAlwaysAs="CONCAT(first_name,' ',last_name)") \
    .addColumn("birthday", "DATE") \
    .addColumn("YOB", "INT", generatedAlwaysAs="YEAR(birthday)") \
    .execute() 
# IdentityGenerator(1,1) = starting value = 1, increment = 1
```
