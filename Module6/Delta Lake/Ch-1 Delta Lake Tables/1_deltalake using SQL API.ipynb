# CREATE DELTA TABLE

### Create Table + Generated Columns, Identity Column

```sql
CREATE TABLE datalake_jung.default.first_sql_api                -- a Delta table by default.
(
    id BIGINT NOT NULL UNIQUE GENERATED ALWAYS AS IDENTITY (START WITH 100 INCREMENT BY 1),
    salary DECIMAL(18,2) NOT NULL,
    first_name STRING NOT NULL,
    last_name STRING NOT NULL,
    full_name STRING GENERATED ALWAYS AS (CONCAT(first_name, ' ', last_name))
)

-- CREATE TABLE datalake_jung.default.first_sql_api                -- explicitly define DELTA format
-- USING DELTA
-- (
--     id INT NOT NULL UNIQUE,
--     salary DECIMAL(18,2) NOT NULL,
--     name STRING NOT NULL
-- )
```
