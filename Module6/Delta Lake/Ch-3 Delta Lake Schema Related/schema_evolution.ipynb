# Scenario
Intentionally writing df2 with a different schema into the existing demo_sink, but now we mergeSchema to demonstrate **schema evolution**.

```sql
%python
# data2 has an incompatible schema
data2 = [
    (1, 100, 'aa', 67),
    (2, 200, 'bb', 36),
    (3, 300, 'cc', 54),
    (4, 400, 'dd', 18),
    (5, 500, 'ee', 20)
]

df2 = spark.createDataFrame(data2, ['id', 'salary', 'name', 'age'])

display(df2)
```

```sql
%python
# Write this new dataframe (this time we use mergeSchema)
df2.write \
    .format("delta") \
    .mode("append") \
    .option("mergeSchema", True) \
    .save("/Volumes/datalake_jung/default/delta_volume/demo_sink/")
```

```sql
%python
# Read this new dataframe 
read_delta = spark.read \
                .format("delta") \
                .load("/Volumes/datalake_jung/default/delta_volume/demo_sink/")

display(read_delta)
```
