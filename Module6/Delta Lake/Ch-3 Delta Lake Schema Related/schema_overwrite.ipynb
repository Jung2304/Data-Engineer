```sql
%python
# data2 has an incompatible schema
data2 = [
    (1, 100, 'aa', 67, 331231),
    (2, 200, 'bb', 36, 3131),
    (3, 300, 'cc', 54, 543543),
    (4, 400, 'dd', 18,2311),
    (5, 500, 'ee', 20,32131)
]

df2 = spark.createDataFrame(data2, ['id', 'salary', 'name', 'age', 'random'])

display(df2)
```

```sql
%python
# Overwrite both the schema and data in demo_sink volumne
df2.write \
    .format("delta") \
    .option("overwriteSchema", True) \
    .mode("overwrite") \
    .save("/Volumes/datalake_jung/default/delta_volume/demo_sink/")
```

```sql
%python
read_delta = spark.read \
                .format("delta") \
                .load("/Volumes/datalake_jung/default/delta_volume/demo_sink/")

display(read_delta)
```
