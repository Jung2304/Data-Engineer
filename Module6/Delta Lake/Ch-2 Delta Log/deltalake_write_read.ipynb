```sql
%python
data = [
    (1, 100, 'aa'),
    (2, 200, 'bb'),
    (3, 300, 'cc'),
    (4, 400, 'dd'),
    (5, 500, 'ee')
]

df = spark.createDataFrame(data, ['id', 'salary', 'name'])

display(df)
```

### Write dataframe as Delta file into directory (Path-based Delta)

```sql
%python
df.write \
    .format("delta") \
    .mode("append") \
    .save("/Volumes/datalake_jung/default/delta_volume/demo_sink/") 
```

### Read using Delta path

```sql
%python
read_delta = spark.read \
        .format("delta") \
        .load("/Volumes/datalake_jung/default/delta_volume/demo_sink/") 

display(read_delta)
```
