## UPSERT with DELTA API

```sql
%python
from delta.tables import DeltaTable
```

```sql
%python
# Create delta object on destination table
dlt_object = DeltaTable.forPath(spark, "/Volumes/datalake_jung/default/delta_volume/dml_sink/")

dlt_object.alias("target").merge(df.alias("source"), "target.id = source.id") \
    .whenMatchedUpdateAll() \
    .whenNotMatchedInsertAll() \
    .execute()
```
