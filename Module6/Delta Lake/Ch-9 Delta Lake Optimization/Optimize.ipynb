## OPTIMIZE
Improves Delta table file layout by compacting small files, repeated writes can produce many small files -> It can improve query performance, especially for large tables.

```sql
%python
df = (spark.read.table("datalake_jung.default.deepclone_table"))

display(df)
```

```sql
%python
df.write \
    .format("delta") \
    .mode("append") \
    .save("/Volumes/datalake_jung/default/delta_volume/optimization/")
```

```sql
OPTIMIZE delta.`/Volumes/datalake_jung/default/delta_volume/optimization/`
```

```sql
DESCRIBE HISTORY delta.`/Volumes/datalake_jung/default/delta_volume/optimization/`
```

## ZORDERING
Reorganize data inside files so that rows with similar values are stored closer together. Data clustering for faster queries.  
--> DATA SKIPPING

```sql
SELECT *
FROM delta.`/Volumes/datalake_jung/default/delta_volume/optimization/`
WHERE id = 1
-- After optimization, took around 1.842s to query data
```

```sql
OPTIMIZE delta.`/Volumes/datalake_jung/default/delta_volume/optimization/`
ZORDER BY (id)
```

```sql
SELECT *
FROM delta.`/Volumes/datalake_jung/default/delta_volume/optimization/`
WHERE id = 1
-- After optimization, took around 1.842s to query data
```
